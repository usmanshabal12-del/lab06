```python
import numpy as np

# 1. Define Input Matrix and Kernel for Convolution
matrix1 = np.array([
    [3, 1, 0, 2, 1],
    [2, 0, 1, 3, 0],
    [4, 1, 0, 1, 2],
    [1, 3, 2, 0, 1],
    [0, 2, 1, 4, 2]
])

kernel = np.array([
    [1,  0, -1],
    [1,  0, -1],
    [1,  0, -1]
])

# 2. Perform Convolution (3x3 sliding window)
output = np.zeros((3, 3), dtype=int)
for i in range(3):
    for j in range(3):
        window = matrix1[i:i+3, j:j+3]
        output[i, j] = np.sum(window * kernel)

print("Convolution Output Matrix:")
print(output)

# 3. Perform 2x2 Max Pooling
pool_size = 2
rows = output.shape[0] - pool_size + 1
cols = output.shape[1] - pool_size + 1
max_output = np.zeros((rows, cols), dtype=int)

for i in range(rows):
    for j in range(cols):
        window = output[i:i+pool_size, j:j+pool_size]
        max_output[i, j] = np.max(window)

print("\nFinal Max Pooling Output:")
print(max_output)

# 4. Define Class Weights and Evaluate Prediction
cat = np.array([0.3, 0.1, 0.2, 0.1])
dog = np.array([0.1, 0.2, 0.1, 0.3])
bird = np.array([0.0, 0.4, 0.1, 0.2])

cat_result = np.sum(cat.reshape(2, 2) * max_output)
dog_result = np.sum(dog.reshape(2, 2) * max_output)
bird_result = np.sum(bird.reshape(2, 2) * max_output)

print("\nClassification Scores:")
print(f"Cat result: {cat_result}")
print(f"Dog result: {dog_result}")
print(f"Bird result: {bird_result}")

results = {
    "Cat": cat_result,
    "Dog": dog_result,
    "Bird": bird_result
}

print(f"\nPredicted Class: {max(results, key=results.get)}")
```
